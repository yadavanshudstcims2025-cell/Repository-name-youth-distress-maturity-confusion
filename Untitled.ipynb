```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import classification_report, confusion_matrix, roc_auc_score
from sklearn.cluster import KMeans

df = pd.read_csv("responses.csv") 
print(df.shape)
print(df.head())
print(df.info())
```

```python
# After you subset to 20–29:
df_20s = df[(df["Age"] >= 20) & (df["Age"] <= 29)].copy()

# Choose a base set of variables to keep
cols_keep = [
    "Age",
    "Gender",
    "Happiness in life",
    "Education",
    "Prioritising workload",
    "Thinking ahead",
    "Final judgement",
    "Reliability",
    "Keeping promises",
    "Punctuality",
    "Achievements",
    "Responding to a serious letter",
    "Assertiveness",
    "Decision making",
    "Judgment calls",
    "Loneliness",
    "Mood swings",
    "New environment",
    "Life struggles",
    "Loss of interest",
    "Energy levels",
    "Health",
    "Internet usage",
    "Finances"
]

df_20s = df_20s[cols_keep]
print(df_20s.head())
print(df_20s.info())
```

```python
# Maturity score: responsibility & planning
maturity_items = [
    "Prioritising workload",
    "Thinking ahead",
    "Final judgement",
    "Reliability",
    "Keeping promises",
    "Punctuality",
    "Achievements",
    "Responding to a serious letter",
    "Assertiveness"
]

# Confusion score: indecision & instability
confusion_items = [
    "Decision making",
    "Judgment calls",
    "Mood swings",
    "New environment",
    "Life struggles"
]

# Distress score: negative emotions & low well-being
distress_items = [
    "Loss of interest",
    "Loneliness",
    "Life struggles",
    "Happiness in life",   # remember: lower = more distress
    "Energy levels",       # lower = more distress
    "Health"               # depending on coding, poorer health = more distress
]
print(df_20s.dtypes)
```

```python
df_20s["Gender_num"] = df_20s["Gender"].map({"male": 0, "female": 1})
df_20s = pd.get_dummies(df_20s, columns=["Education"], drop_first=True)
```

```python
print(df_20s["Punctuality"].unique())
print(df_20s["Internet usage"].unique())
```

```python
punct_map = {
    "i am often running late": 1,
    "i am always on time": 2,
    "i am often early": 3
}

df_20s["Punctuality_num"] = df_20s["Punctuality"].map(punct_map)
internet_map = {
    "no time at all": 0,
    "less than an hour a day": 1,
    "few hours a day": 2,
    "most of the day": 3
}

df_20s["Internet_usage_num"] = df_20s["Internet usage"].map(internet_map)
```

```python
maturity_items = [
    "Prioritising workload",
    "Thinking ahead",
    "Final judgement",
    "Reliability",
    "Keeping promises",
    "Achievements",
    "Responding to a serious letter",
    "Assertiveness",
    "Punctuality_num"   # use the numeric column
]

confusion_items = [
    "Decision making",
    "Judgment calls",
    "Mood swings",
    "New environment",
    "Life struggles"
]

distress_items = [
    "Loss of interest",
    "Loneliness",
    "Life struggles",
    "Happiness in life",
    "Energy levels",
    "Health"
]

df_20s["Maturity_score"]  = df_20s[maturity_items].mean(axis=1)
df_20s["Confusion_score"] = df_20s[confusion_items].mean(axis=1)
df_20s["Distress_score"]  = df_20s[distress_items].mean(axis=1)
```

```python
import numpy as np

# Use median as a simple threshold
threshold = df_20s["Distress_score"].median()
df_20s["High_distress"] = (df_20s["Distress_score"] >= threshold).astype(int)

df_20s["High_distress"].value_counts()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Check distributions
for col in ["Maturity_score", "Confusion_score", "Distress_score"]:
    sns.histplot(df_20s[col], kde=True)
    plt.title(f"Distribution of {col}")
    plt.show()

# Correlation between scores
scores = df_20s[["Maturity_score", "Confusion_score", "Distress_score"]]
sns.heatmap(scores.corr(), annot=True, cmap="coolwarm")
plt.title("Correlation among maturity, confusion, distress")
plt.show()
```

```python
import statsmodels.api as sm

X = df_20s[["Maturity_score", "Confusion_score", "Age", "Gender_num"]]
y = df_20s["Distress_score"]

X = sm.add_constant(X)
model = sm.OLS(y, X).fit()
print(model.summary())
```

```python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import classification_report, confusion_matrix, roc_auc_score

feature_cols = [
    "Maturity_score",
    "Confusion_score",
    "Age",
    "Gender_num",
    "Internet_usage_num"
]

X = df_20s[feature_cols]
y = df_20s["High_distress"]

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=42
)

scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

clf = LogisticRegression()
clf.fit(X_train_scaled, y_train)

y_pred = clf.predict(X_test_scaled)
y_proba = clf.predict_proba(X_test_scaled)[:, 1]

print(classification_report(y_test, y_pred))
print("ROC-AUC:", roc_auc_score(y_test, y_proba))
print("Confusion matrix:", confusion_matrix(y_test, y_pred))
```

```python
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler

cluster_features = df_20s[["Maturity_score", "Confusion_score", "Distress_score"]]
cluster_scaled = StandardScaler().fit_transform(cluster_features)

kmeans = KMeans(n_clusters=3, random_state=42)
df_20s["Cluster"] = kmeans.fit_predict(cluster_scaled)

# Look at cluster means
print(df_20s.groupby("Cluster")[["Maturity_score", "Confusion_score", "Distress_score"]].mean())

# Simple visualization
sns.scatterplot(
    data=df_20s, x="Maturity_score", y="Distress_score",
    hue="Cluster", palette="Set2"
)
plt.title("Clusters of youth (20s) by maturity vs distress")
plt.show()
```

```python
from scipy import stats
import numpy as np
import pandas as pd

# Clean subset for gender test
data_h3 = df_20s[["Distress_score", "Gender_num"]].replace([np.inf, -np.inf], np.nan)
data_h3 = data_h3.dropna()

male_distress   = data_h3.loc[data_h3["Gender_num"] == 0, "Distress_score"]
female_distress = data_h3.loc[data_h3["Gender_num"] == 1, "Distress_score"]

# Two-sample t-test (assuming unequal variances)
t_stat, p_val = stats.ttest_ind(male_distress, female_distress, equal_var=False)
print("t-statistic:", t_stat)
print("p-value:", p_val)
print("Male mean:", male_distress.mean())
print("Female mean:", female_distress.mean())
```

```python
from scipy import stats

data_h4 = df_20s[["Distress_score", "Internet_usage_num"]].replace([np.inf, -np.inf], np.nan)
data_h4 = data_h4.dropna()

groups = []
for level in sorted(data_h4["Internet_usage_num"].unique()):
    groups.append(data_h4.loc[data_h4["Internet_usage_num"] == level, "Distress_score"])

f_stat, p_val = stats.f_oneway(*groups)
print("F-statistic:", f_stat)
print("p-value:", p_val)

# Check group means for interpretation
group_means = data_h4.groupby("Internet_usage_num")["Distress_score"].mean()
print(group_means)
```
